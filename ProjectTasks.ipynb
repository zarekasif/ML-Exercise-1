## 🧩 Overview

### Goal
Build baseline classification pipelines on multiple datasets,  
evaluate with both **Holdout** and **Cross Validation**,  
and analyze performance, preprocessing effects, and runtime.

### Datasets
- `adult`
- `cancer`
- `loan`
- `wine`

### Models
- Logistic Regression  
- Random Forest  
- SVM (RBF)

---

## 🪜 Task List

### 1. Data Loading
- Load all datasets (Adult, Cancer, Loan, Wine)
- Separate `X` (features) and `y` (target)

### 2. Data Exploration & Preprocessing
- Check missing values, outliers, and feature scales  
- Build preprocessing pipelines:
  - **Numerical**: `SimpleImputer(strategy="median")` + `StandardScaler()`
  - **Categorical**: `SimpleImputer(strategy="most_frequent")` + `OneHotEncoder()`
- Confirm that missing value handling works (✅ done)

### 3. Classification (Baseline)
- Use three classifiers: Logistic Regression, Random Forest, SVM (RBF)
- Apply to each dataset using the baseline preprocessing pipeline
- Evaluate on **holdout split** (e.g., 80/20)
- Save per-dataset CSV results:  
  `adult_baseline.csv`, `cancer_baseline.csv`, `loan_baseline.csv`, `wine_baseline.csv`

### 4. Evaluation Metrics
- Compute multiple performance metrics:
  - Accuracy  
  - Balanced Accuracy  
  - Macro F1
- Save metrics and runtime in:
  - `runtime.csv`
  - `{dataset}_baseline.csv`

### 5. Holdout vs Cross-Validation
- Implement **5-fold Stratified Cross Validation**
- Compare mean scores with holdout results
- Save CV results as `cv.csv`

### 6. Ablation Study (Preprocessing Effects)
- Compare **SVM with and without scaling**
- Save ablation results as `ablation.csv`

### 7. Result Aggregation
- Collect all CSV files in `/results`:
  - `cv.csv`
  - `ablation.csv`
  - `*_baseline.csv`
  - `runtime.csv`

### 8. Best Summary Table (Main Output)
- Select the **best-performing model per dataset** (based on CV F1)
- Merge holdout metrics and runtime information
- Save final table:  
  `results/summary/summary_best_4x3.csv`  
  (or enriched version: `summary_best_4x3_enriched.csv`)
